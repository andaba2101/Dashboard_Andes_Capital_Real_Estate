# 📊 Proyecto: Análisis Comercial Inmobiliario

El sector inmobiliario necesita evaluar su **desempeño comercial** para comprender su crecimiento, rentabilidad y comportamiento de clientes.

Se solicita construir un **dashboard ejecutivo** que permita analizar ventas, clientes y propiedades para apoyar decisiones estratégicas basadas en datos.


# 📂 Descripción de los Datasets

🔹 **hecho_ventas_propiedades**  
Tabla de hechos con las transacciones de venta (precio, cliente, propiedad, canal, fecha).

**Cohortes de clientes**

## Insights accionables

El segmento de clientes de primera vez concentra una parte importante de las compras, en todos los tipos de propiedades. En el caso de las casas, este segmento representa aproximadamente $1.379 millones, mientras que los inversionistas aportan cerca de $852 millones y los clientes de alto patrimonio alrededor de $328 millones. Esto sugiere que existe una oportunidad relevante para captar compradores que ingresan por primera vez al mercado inmobiliario o buscan su primera vivienda. Es de entender que el mercado inmobiliario no es tan flexible, puesto que en practicamente todos los casos son inversiones a largo plazo (construcción de patrimonio) por lo que los ingresos representados en forma de inversión debería estudiarse a mayor detalle ya que pueden provenir de captación por participación inmobiliaria o como segundo inmueble para muchos.

Las cohortes iniciales, especialmente las correspondientes a enero, febrero, marzo y abril de 2023, muestran los mayores volúmenes de clientes y una recurrencia más amplia a través del tiempo. La cohorte de marzo de 2023 registra el mayor total, con 1.202 clientes, seguida de abril, con 1.085, y febrero, con 782. En contraste, las cohortes más recientes presentan volúmenes menores. Las cohorte mencionadas no presentaban una tendencia per se por lo que convendría indagar que sucedió en ese periodo que incentivo arduamente la compra de inmuebles.

Las ventas presentan un comportamiento fluctuante durante el periodo. Se observan repuntes en algunos meses y descensos marcados en otros, especialmente alrededor de los meses de menor actividad. Por esta razón, el negocio debería analizar la estacionalidad y establecer campañas comerciales diferenciadas para los periodos de baja demanda. En general se aprecia mayor compra de inmuebles cuando el mercado observa bajas en los precios, lo que indica alta sensibilidad.

Los precios de los inmuebles muestran una tendencia general al alza, aunque con variaciones mensuales significativas. El crecimiento de los precios alcanza valores elevados en algunos meses, mientras que en otros se registran disminuciones. Esta situación puede estar relacionada con cambios en el tipo de propiedades vendidas, la ciudad, el canal de comercialización o el perfil del comprador.

Este segmento puede utilizarse para aumentar el volumen comercial y atraer compradores de primera vez.

- Diseñar programas de retención y recompra, especialmente para clientes de las cohortes iniciales que ya han demostrado mayor permanencia y recurrencia. Estos programas pueden incluir beneficios para compradores recurrentes, incentivos por recomendación y ofertas personalizadas.

- Implementar campañas específicas para clientes de primera vez, debido a su importante participación en los ingresos. La estrategia puede incluir asesoría financiera, simuladores de compra, acompañamiento durante el proceso y productos de entrada con precios accesibles.

- Monitorear periódicamente el precio promedio y el ticket promedio, con el fin de determinar si el aumento de ingresos proviene de un mayor volumen de ventas, de precios más altos o de una combinación de ambos factores.
